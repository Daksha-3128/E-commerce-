```python
df.dropna(inplace=True)
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.head(10)
```

```python
df['engagement']=df['time_on_site']*df['pages_viewed']
```

```python
df

```

```python
df['avg_time_per_page']=df['time_on_site']/df['pages_viewed']
```

```python
df['avg_time_per_page']
```

```python
import numpy as np

df['returning_customer']=np.where(df['previous_purchases']>0,"Yes","No")
```

```python
df[['previous_purchases','returning_customer']].head(10)
```

```python
df['age_group']=np.where(df['age']<25,'Young',
                         np.where(df['age']<40,'Adult',
                         np.where(df['age']<60,'Middle-aged',
                         'Senior')))
```

```python
df[['age','age_group']].head(20)
```

```python
df.to_csv("cleaned_ecommerce_data.csv",index=False)
```
